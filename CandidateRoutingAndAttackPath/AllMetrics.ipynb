# AllMetrics разделен на самостоятельные ноутбуки

Запускайте этапы по порядку:

1. [`01_TargetDefinitionAndMetrics.ipynb`](01_TargetDefinitionAndMetrics.ipynb) — target-aware label, audit разметки, исторические метрики и candidate routing;
2. [`02_BalancedCausalPath.ipynb`](02_BalancedCausalPath.ipynb) — matched quartets и signed causal-path decomposition;
3. [`03_CausalRepair.ipynb`](03_CausalRepair.ipynb) — sign-selective intervention/repair;
4. [`04_CausalTransplant.ipynb`](04_CausalTransplant.ipynb) — fixed-cell sufficiency-test;
5. [`05_TargetCandidateSet.ipynb`](05_TargetCandidateSet.ipynb) — candidate handoff и causal objective для множества взаимозаменяемых target candidates.

Каждый ноутбук запускается с чистого kernel. Все тяжелые этапы resumable и переиспользуют существующие SQLite/cache-файлы.

Старый монолит вместе с сохраненными outputs находится в [`AllMetrics_legacy.ipynb`](AllMetrics_legacy.ipynb). Он больше не является точкой запуска.
