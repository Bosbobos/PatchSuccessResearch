### Что означают метрики success/fail

Ниже описаны метрики, которые считаются в `run_success_failure_metrics()`. Везде `Δ = F_l(x_patch) - F_l(x_clean)` — изменение feature map выбранного слоя `l`, вызванное патчем. Если `Δ` имеет форму `[C, H, W]`, то пространственная карта энергии считается как `E[h,w] = sqrt(mean_c Δ[c,h,w]^2)`.

**`delta_l2_rms`**

Формула: `sqrt(mean_i E_i^2)`, где `i` пробегает все spatial-позиции карты энергии.

Суть: общий RMS-масштаб изменения фичей. Чем выше значение, тем сильнее патч в среднем меняет представление слоя.

**`delta_abs_mean`**

Формула: `mean_i |E_i|`. Так как `E_i >= 0`, практически это средняя энергия изменения по spatial-карте.

Суть: средняя интенсивность расползания патча по слою без акцента на экстремальные пики.

**`delta_max`**

Формула: `max_i E_i`.

Суть: максимальная локальная сила изменения. Полезно, когда патч создает очень сильный, но узкий всплеск в feature map.

**`delta_gini`**

Формула: `Gini(E) = sum_i (2i - n - 1) E_(i) / (n * sum_i E_i)`, где `E_(i)` — отсортированные по возрастанию значения энергии.

Суть: концентрация изменения. Значение ближе к `0` означает, что патч расползается равномерно; значение ближе к `1` означает, что изменение сосредоточено в малой части карты.

**`roi_energy_frac`**

Формула: `sum_{i in ROI_patch} E_i / sum_i E_i`. ROI строится проекцией bbox патча на spatial-grid слоя.

Суть: доля энергии изменения, оставшаяся внутри области патча. Низкое значение при высокой общей энергии означает, что влияние патча ушло за пределы самого патча.

**`topk_energy_frac`**

Формула: `sum_{i in TopK(E)} E_i / sum_i E_i`, где по умолчанию `K = 5%` spatial-позиций.

Суть: насколько изменение сконцентрировано в самых сильных spatial-позициях. Это дополняет `delta_gini`, но проще интерпретируется как доля энергии в top-k областях.

**`align_cosine`**

Формула: `cos(|Δ|, |A|) = <|Δ|, |A|> / (||Δ||_2 ||A||_2)`, где `A` — карта важности `SegmentIG[0;0.1]`, распрямленная в тот же вектор, что и `Δ`.

Суть: глобальная похожесть карты атакованных фич и карты важных фич. Чем выше значение, тем сильнее патч меняет именно те нейроны/позиции, которые SegmentIG считает важными для target.

**`align_top_jaccard`**

Формула: `|TopK(|Δ|) ∩ TopK(|A|)| / |TopK(|Δ|) ∪ TopK(|A|)|`, где `K = 5%` элементов.

Суть: совпадение самых атакованных и самых важных фичей как множеств. Метрика игнорирует масштаб значений и смотрит только на пересечение top-k.

**`importance_energy_in_delta_top`**

Формула: `sum_{i in TopK(|Δ|)} |A_i| / sum_i |A_i|`.

Суть: какая доля важности SegmentIG попала в области, которые патч изменил сильнее всего. Высокое значение означает, что top-delta области действительно важны для target.

**`delta_energy_in_importance_top`**

Формула: `sum_{i in TopK(|A|)} |Δ_i| / sum_i |Δ_i|`.

Суть: какая доля энергии атаки пришлась на самые важные по SegmentIG области. Высокое значение означает, что патч направленно бьет по важным фичам, а не просто сильно меняет слой где-то еще.

**`roc_auc` и `best_accuracy` в таблице качества**

Для каждой метрики строится бинарный классификатор success/fail по одному порогу. `roc_auc` показывает качество ранжирования успешных атак выше или ниже неуспешных. `best_accuracy` — максимальная accuracy по всем порогам и двум направлениям (`metric >= threshold` или `metric <= threshold`).


```python
# Сравнение delta_energy_in_importance_top для Full IG, SegmentIG и NAA
rows_df = pd.DataFrame(sf['rows'])
quality_df = pd.DataFrame(sf['quality']).sort_values('roc_auc', ascending=False)
energy_metrics = [
    'full_ig_delta_energy_in_importance_top',
    'segmentig_delta_energy_in_importance_top',
    'naa_delta_energy_in_importance_top',
]
missing = [m for m in energy_metrics if m not in rows_df.columns or m not in set(quality_df['metric'])]
if missing:
    raise RuntimeError(f'Нет метрик {missing}. Перезапусти предыдущую ячейку с run_success_failure_metrics(), чтобы пересчитать sf.')
display(rows_df[['path', 'success', 'drop'] + energy_metrics].head())
energy_quality = quality_df[quality_df['metric'].isin(energy_metrics)].set_index('metric').loc[energy_metrics].reset_index()
display(energy_quality)
for _, row in energy_quality.iterrows():
    print(row['metric'])
    display(Image(filename=row['figure_path']))
```
